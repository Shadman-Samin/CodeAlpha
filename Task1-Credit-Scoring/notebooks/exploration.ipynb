# Credit Scoring — Exploration
German Credit (1000 rows). Run `python src/train.py` for full training.

```python
import pandas as pd
df = pd.read_csv('../data/german_credit.csv')
print(df.shape)
display(df['target'].value_counts(normalize=True))
display(df.describe(include='all').T.head(25))
```

```python
import seaborn as sns, matplotlib.pyplot as plt
fig, ax = plt.subplots(1,2, figsize=(10,4))
sns.histplot(df['age'], kde=True, ax=ax[0]); ax[0].set_title('Age')
sns.histplot(df['credit_amount'], kde=True, ax=ax[1]); ax[1].set_title('Credit amount')
plt.tight_layout(); plt.show()
```
